## 5. 速度指令によるベース制御 / Base velocity control

**JP:** `move_base_vel(vx, vy, vw)` はベースの速度指令を設定します。

- `vx`: 前進速度 [m/s]（正 = 前進）
- `vy`: 横方向速度 [m/s]（正 = 左）
- `vw`: 回転速度 [deg/s]（正 = 反時計回り）

この指令は変更するか `stop_base()` を呼ぶまで**持続**します。

**EN:** `move_base_vel(vx, vy, vw)` sets the base velocity command.

- `vx`: Forward velocity [m/s] (positive = forward)
- `vy`: Lateral velocity [m/s] (positive = left)
- `vw`: Rotational velocity [deg/s] (positive = counter-clockwise)

The command **persists** until you change it or call `stop_base()`.

まず前進させてみましょう / Let's move forward first:

```python
# 前進 / Move forward at 0.1 m/s for 2 seconds
move_base_vel(0.1, 0, 0)
run(2.0)
show_video()
```

## 6. 回転 / Rotation

**JP:** 次にその場で回転させます。`vw` に正の値を与えると反時計回りに回転します。

**EN:** Now let's rotate in place. A positive `vw` rotates counter-clockwise.

```python
# 反時計回りに回転 / Rotate counter-clockwise at 10 deg/s for 2 seconds
move_base_vel(0, 0, 10)
run(2.0)
show_video()
```

## 7. 停止 / Stop the base

**JP:** `stop_base()` を呼ぶと速度指令をゼロにしてベースを停止します。

**EN:** Call `stop_base()` to zero the velocity command and stop the base.

```python
# 停止 / Stop the base
stop_base()
run(0.5)
show_video()
```

## 8. 目標位置へのナビゲーション / Goal-based navigation

**JP:** `move_base_goal(x, y, theta, duration)` は目標位置と姿勢を指定して
ベースを移動させます。軌跡が自動生成され、指定した時間内で目標に到達します。

- `x`, `y`: 目標位置 [m]
- `theta`: 目標姿勢 [deg]
- `duration`: 目標到達までの時間 [s]（戻り値としても返されます）

**EN:** `move_base_goal(x, y, theta, duration)` commands the base to move
to a target pose. A trajectory is automatically generated to reach the
goal within the specified duration.

- `x`, `y`: Target position [m]
- `theta`: Target heading [deg]
- `duration`: Time to reach the goal [s] (also returned as the return value)

```python
# 目標位置へ移動 / Move to goal (x=0.5, y=0.3, theta=45 deg)
move_base_goal(0.5, 0.3, 45)
run(3.0)
show_video()
```

## 9. 現在位置の確認 / Check the current position

**JP:** `get_base_pos()` は現在のベース位置を `(x, y, yaw_deg)` として返します。

**EN:** `get_base_pos()` returns the current base position as `(x, y, yaw_deg)`.

```python
# 現在位置を確認 / Check current base position
print(get_base_pos())
```

## 10. 原点への復帰 / Return to the origin

**JP:** 最後に原点 `(0, 0, 0)` に戻ります。

**EN:** Finally, return to the origin `(0, 0, 0)`.

```python
# 原点へ復帰 / Return to the origin
move_base_goal(0, 0, 0)
run(3.0)
show_video()
```
