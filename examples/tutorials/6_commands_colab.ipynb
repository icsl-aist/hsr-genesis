問題が発生した場合は[トラブルシューティングノートブック](7_troubleshoot_colab.ipynb)を参照してください。


```python
import importlib, urllib.request

# Fetch standalone bootstrap from GitHub (zero hsr_genesis imports).
exec(urllib.request.urlopen(
    "https://raw.githubusercontent.com/icsl-aist/hsr-genesis/main/colab_setup.py"
).read())

# One-call setup: installs deps, clones repo (with submodules),
# configures EGL for headless GPU rendering. Safe to re-run.
# See the troubleshoot notebook if anything goes wrong.
setup_colab()
```

```python
from hsr_genesis.tutorial_utils import *

init_sim()
```

ベースの初期位置へ移動させる

Move the moving base to the initial position:

```python
move_base_goal(0, 0, 0)
run(3.0)
show_video()
```

アームをニュートラル姿勢へ移動させる

Move the arm to the neutral posture:

```python
move_arm_neutral()
run(2.0)
show_video()
```

アームの初期姿勢へ移動させる

Move the arm to the initial posture:

```python
move_arm_init()
run(2.0)
show_video()
```

ハンドを開く

Open the hand:

```python
move_hand(1.0)
run(1.0)
show_video()
```

ハンドを閉じる

Close the hand:

```python
move_hand(0.0)
run(1.0)
show_video()
```

現在のベース位置を取得する

Get the current base position:

```python
print(get_base_pos())
```

現在のハンド位置を取得する

Get the current hand position:

```python
print(get_hand_pos())
```

箱を配置する

Spawn a box:

```python
box = spawn_box((0.5, 0.0, 0.02))
run(0.5)
show_video()
```

球を配置する

Spawn a sphere:

```python
sphere = spawn_sphere((0.5, 0.2, 0.03))
run(0.5)
show_video()
```

円柱を配置する

Spawn a cylinder:

```python
cyl = spawn_cylinder((0.5, -0.2, 0.1))
run(0.5)
show_video()
```

フレームバッファをクリアする

動画用に保存されたフレームをすべて削除し、新しく動画の撮影を始められます。

Clear the frame buffer

This removes all captured frames so you can start a fresh video segment.

```python
clear_frames()
```

動画を保存する

取得したフレームをMP4ファイルとして保存します。

Save the video

Save the captured frames to an MP4 file.

```python
save_video('/content/my_video.mp4')
```

シミュレーションをリセットする

現在のシーンを破棄し、最初から再構築します。
