# HSR Genesis Tutorial Series

A progressive, hands-on tutorial series for learning to simulate and control the
**Toyota Human Support Robot (HSR)** with the [Genesis](https://genesis-embodied-ai.github.io/) physics engine
and the [hsr-genesis](https://github.com/icsl-aist/hsr-genesis) package.

Each notebook runs entirely in **Google Colab** with a free T4 GPU — no local
setup required. Headless rendering is handled by `mediapy` + an offscreen
Genesis camera, so every step produces an inline video you can watch right in
the notebook.

The series is inspired by the bilingual
[hsr-project/notebooks](https://github.com/hsr-project/notebooks) and uses the
simplified `hsr_genesis.tutorial_utils` API so you can focus on *what* the robot
does rather than on Genesis internals.

---

## チュートリアル一覧 / Tutorial List

Run the notebooks in order — each one builds on the previous.


### 1. Basics — `1_basics_colab.ipynb`

Install Genesis, clone the repo, initialize a headless HSR scene, and take your
first simulation steps. You will spawn a box and watch the robot standing in
its environment.

基礎編 — Genesis のインストール、リポジトリのクローン、ヘッドレス HSR
シーンの初期化、最初のシミュレーションステップを実行します。ボックスを
配置し、ロボットが環境に立っている様子を確認します。

[Open in Colab](https://colab.research.google.com/github/icsl-aist/hsr-genesis/blob/main/examples/tutorials/1_basics_colab.ipynb)

### 2. Base Control — `2_base_control_colab.ipynb`

Drive the omni-directional base with velocity commands (`move_base_vel`) and
navigation goals (`move_base_goal`). Inspect the robot pose with `get_base_pos`.

移動制御編 — 全方向移動台車を速度指令 (`move_base_vel`) とナビゲーション
目標 (`move_base_goal`) で操作します。`get_base_pos` で姿勢を確認します。

[Open in Colab](https://colab.research.google.com/github/icsl-aist/hsr-genesis/blob/main/examples/tutorials/2_base_control_colab.ipynb)

### 3. Arm Control — `3_arm_control_colab.ipynb`

Move the arm to named poses (`move_arm_neutral`, `move_arm_init`) and to
arbitrary joint angles (`move_arm_joints`). Visualize the end-effector with
`forward_kinematics` and reach targets with `move_wholebody_ik`.

腕制御編 — 腕を名前付き姿勢 (`move_arm_neutral`, `move_arm_init`) や任意の
関節角度 (`move_arm_joints`) に動かします。`forward_kinematics` で
手先の位置を可視化し、`move_wholebody_ik` で目標に到達します。

[Open in Colab](https://colab.research.google.com/github/icsl-aist/hsr-genesis/blob/main/examples/tutorials/3_arm_control_colab.ipynb)

### 4. Gripper Control — `4_gripper_colab.ipynb`

Open and close the hand (`move_hand`), grasp objects with force control
(`grasp_object`), and release them (`release_object`). Practice picking up
a spawned box.

把持編 — ハンドの開閉 (`move_hand`)、力制御による把持 (`grasp_object`)、
解除 (`release_object`) を学びます。配置したボックスの把持を練習します。

[Open in Colab](https://colab.research.google.com/github/icsl-aist/hsr-genesis/blob/main/examples/tutorials/4_gripper_colab.ipynb)

### 5. Full Grasp Sequence — `5_grasp_colab.ipynb`

Put it all together: navigate to an object, reach with whole-body IK, grasp,
transport, and release. A complete pick-and-place task using the full
`tutorial_utils` API.

統合編 — これまでの要素を組み合わせます。対象物へ移動し、全身逆運動学
で到達し、把持して運搬・解除する、ピック＆プレースの完全なタスクを
`tutorial_utils` API の全機能を使って実行します。

[Open in Colab](https://colab.research.google.com/github/icsl-aist/hsr-genesis/blob/main/examples/tutorials/5_grasp_colab.ipynb)

### 6. Commands Reference — `6_commands_colab.ipynb`

A quick-reference glossary of all `tutorial_utils` functions with live demos.
Handy when you forget a function name or want to see a specific command in action.

コマンド集 — `tutorial_utils` の全関数を実演付きで紹介する
早見表です。関数名を忘れた時や特定のコマンドの動作を確認したい時に便利です。

[Open in Colab](https://colab.research.google.com/github/icsl-aist/hsr-genesis/blob/main/examples/tutorials/6_commands_colab.ipynb)

### 7. Troubleshooting — `7_troubleshoot_colab.ipynb`

Diagnose and fix common issues: GPU not detected, import errors, missing meshes,
EGL rendering problems, OOM, and more. Includes a one-click diagnostic cell and
manual setup instructions.

トラブルシューティング — GPU 未検出、インポートエラー、メッシュ不足、
EGL レンダリング問題、OOM などの一般的な問題の診断と解決。ワンクリック診断セルと
手動セットアップ手順を含みます。

[Open in Colab](https://colab.research.google.com/github/icsl-aist/hsr-genesis/blob/main/examples/tutorials/7_troubleshoot_colab.ipynb)

### 8. Parallel Simulation on GPU — `8_parallel_simulation_colab.ipynb`

Bridge between single-env control (tutorials 1–7) and GPU batched parallel sim used in RL notebooks. Builds N parallel environments, shows the one-to-one correspondence between single-env API and `*_batched` controller methods, and benchmarks batched `scene.step()` vs N single-env calls.

GPU 並列シミュレーションの導入。単一環境制御(チュートリアル 1–7)と RL ノートブック(9–10)の橋渡し。N 個の並列環境を構築し、単一環境 API と `*_batched` メソッドの対応関係を示し、バッチ化 `scene.step()` のベンチマークを計測します。

[Open in Colab](https://colab.research.google.com/github/icsl-aist/hsr-genesis/blob/main/examples/tutorials/8_parallel_simulation_colab.ipynb)

### 9. Grasp Learning with CMA-ES — `9_grasp_learning_cmaes_colab.ipynb`

Learn to optimize grasp parameters (pre-grasp height, grasp offset, gripper
effort, hold time) for each YCB object using CMA-ES evolution strategy. Combines
EvoTorch's GPU-accelerated optimizer with Genesis parallel simulation to search
a 28D parameter space across 7 objects.

学習編 — CMA-ES進化戦略を用いて、各YCB物体の把持パラメータ（把持前高さ、
把持オフセット、グリッパ力、保持時間）を最適化する方法を学びます。
EvoTorchのGPU加速CMA-ESとGenesis並列シミュレーションを組み合わせ、
7物体の28次元パラメータ空間を探索します。

[Open in Colab](https://colab.research.google.com/github/icsl-aist/hsr-genesis/blob/main/examples/tutorials/9_grasp_learning_cmaes_colab.ipynb)

### 10. Grasp Learning with PPO + IK Curriculum — `10_grasp_learning_ppo_colab.ipynb`

Train a **PPO residual policy** that learns 9D corrections (delta arm + delta base +
gripper effort) on top of IK pick trajectories. A 3-stage curriculum gradually shifts
control from pure IK to policy-dominant, using stable-baselines3 PPO with Genesis
parallel simulation. Builds on the CMA-ES optimized parameters from Section 8.

学習編（PPO）— IK把持軌跡の上に9次元の残差補正（腕・台車・グリッパ）を学習する
**PPO残差ポリシー**を訓練します。3段階のカリキュラムでIK主導からポリシー主導へ
移行し、stable-baselines3 PPOとGenesis並列シミュレーションを使用します。
Section 8のCMA-ES最適化パラメータを基盤として使用します。

[Open in Colab](https://colab.research.google.com/github/icsl-aist/hsr-genesis/blob/main/examples/tutorials/10_grasp_learning_ppo_colab.ipynb)
