## 5. What is HSR? / HSR とは？

The **Human Support Robot (HSR)** is a compact, mobile manipulator developed by
Toyota Motor Corporation to assist people in home environments. It features:

- An **omni-directional base** for holonomic movement in tight spaces.
- A **telescoping torso lift** that raises the arm to reach high places.
- A **5-DOF arm** with a wrist flex/roll for manipulation.
- A **gripper hand** with a force-controlled grasping mechanism.
- A **head** with a tilt joint and onboard sensors.

In this tutorial series we simulate the HSR in Genesis, a GPU-based physics
simulation engine, using the `hsr-genesis` package which wraps the robot's URDF,
controllers, and analytic inverse kinematics.

**HSR (Human Support Robot)** はトヨタ自動車が開発した、家庭環境での生活を
支援するための小型移動マニピュレータです。主な特徴:

- 狭い空間でも自在に移動できる **全方向移動台車**。
- 高い場所に届く **伸縮式胴体リフト**。
- 把持のための手首屈曲/回転を備えた **5 自由度アーム**。
- 力制御による把持機構を備えた **ハンド**。
- チルト関節とセンサを搭載した **頭部**。

このチュートリアルシリーズでは、ロボットの URDF・コントローラ・解析的逆運動学を
ラップした `hsr-genesis` パッケージを使い、GPU ベース物理シミュレーションエンジン
Genesis 上で HSR をシミュレーションします。

## 6. First simulation step / 最初のシミュレーションステップ

Let the simulation run for 1 second. `run(seconds)` advances the physics and
captures camera frames; `show_video()` then displays them as an inline video.
You should see the HSR standing still on the ground plane.

シミュレーションを 1 秒間実行します。`run(seconds)` が物理を進めつつカメラ
フレームを取得し、`show_video()` が動画としてその場に表示します。HSR が
地面の上に静止して立っている様子が見えるはずです。

```python
run(1.0)
show_video()
```

## 7. Spawn a box / ボックスを配置する

Now let's add an object to the scene. `spawn_box((x, y, z))` spawns a small box
at the given world position. We place it 0.5 m in front of the robot at ground
level, then run the simulation for another second so the box settles, and
display the result.

次にシーンにオブジェクトを追加します。`spawn_box((x, y, z))` は指定した
ワールド座標に小さなボックスを配置します。ロボットの前方 0.5 m・地面の
高さに配置し、ボックスが落ち着くようもう 1 秒シミュレーションを実行して
結果を表示します。

```python
clear_frames()
spawn_box((0.5, 0.0, 0.02))
run(1.0)
show_video()
```
